```python
from pydantic import BaseModel
```

```python
class User(BaseModel):
    id: int
    name: str
    is_active: bool = True
```

```python
u = User(id= "101", name="KK")
```

```python
print(u)
```

```python
print(type(u), type(u.id), type(u.name), type(u.is_active), u.id)
```

```python
print(u.model_dump())
```

```python
class Product(BaseModel):
    id: int
    price: float
    in_stock: bool 
```

```python
try:
    p = Product(id="A1", price="free", in_stock="yes")
except ValidationError as e:
    print(e)
```

```python
from pydantic import BaseModel, Field, ValidationError
```

```python
class Signup(BaseModel):
    username: str = Field(min_length = 3, max_length = 20)
    age: int = Field(ge=18, le=60)
    score: float = Field(default=0.0, ge=0, le=100)
```

```python
try:
    s = Signup(username="ai", age=15, score=150)
except ValidationError as e:
    print(e)
```

```python
from pydantic import BaseModel
```

```python
class Address(BaseModel):
    city: str
    pincode: int
    
```

```python
class Customer(BaseModel):
    name: str
    address: Address
    
```

```python
payload = {
    "name": "Goftus", 
    "address": {"city": "Chennai", "pincode": "600001"} #string instead of int
}
```

```python
c = Customer(**payload)
```
