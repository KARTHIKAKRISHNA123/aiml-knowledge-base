```python
from google import genai
import os

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

chat = client.chats.create(
    model="gemini-3.8-flash"
)

response = chat.send_message(
    "a+2=7, find a. Explain it like a math teacher."
)

print(response.text)
```

```python
import ollama

response = ollama.chat(
    model="gemma:2b",
    messages=[
        {
            "role": "system",
            "content": "Act like a math teacher"
        },
        {
            "role": "user",
            "content": "a+2=7, find a"
        }
    ]
)

print(response["message"]["content"])
```

```python
from google import genai
import os

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="a+2=7, find a",
    config={
        "system_instruction": "Act like a math teacher"
    }
)

print(response.text)
```
