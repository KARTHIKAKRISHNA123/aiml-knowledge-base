```python
#Melt & Pivot
#Melt - Wide to Long while preserving values

df = pd.DataFrame({
    "country" : ["USA", "USA", "India", "India"],
    "year" : [2020, 2021, 2020, 2021],
    "sales" : [100, 120, 90, 110],
    "profit" : [20, 25, 18, 22]
})
df
```

```python
melted_df = df.melt(
    id_vars = ["country", "year"],
    value_vars = ["sales", "profit"],
    var_name = "metrics",
    value_name = "value of either sales or profit"
)
```

```python
#Melt & Pivot
#Melt - Wide to Long while preserving values

df = pd.DataFrame({
    "country" : ["USA", "USA", "India", "India"],
    "year" : [2020, 2021, 2020, 2021],
    "sales" : [100, 120, 90, 110],
    "profit" : [20, 25, 18, 22],
    "tax" : [10, 20, 30, 40]
})
df
```

```python
df.melt(
    id_vars = ["country", "year"],
    value_vars = ["sales", "profit", "tax"],
    var_name = "metrics",
    value_name = "value of either sales or profit"
)
```

```python
#pivot -long to wide while preserving data opposite of melt
original = melted_df.pivot(
    index=["country", "year"], #unique values
    columns = "metrics",
    values = "value of either sales or profit"
    
    
    
)
original
```

```python
#Merging and Joining Data

df_customers = pd.DataFrame({
    "customer_id" : [1, 2, 3, 4],
    "name" : ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id" : [101, 102, 103, 104],
    "customer_id" : [2, 1, 4, 5],
    "amount" : [250, 120, 300, 180]
})
df_customers

    
    
```

```python
df_orders
```

```python
pd.merge(df_customers, df_orders, on="customer_id") #on represents foreign key in sql
#inner join
```

```python
pd.merge(df_customers, df_orders, on="customer_id", how="left") #left join
```

```python
pd.merge(df_customers, df_orders, on="customer_id", how="right") #right join
```

```python
pd.merge(df_customers, df_orders, on="customer_id", how="outer")
```
