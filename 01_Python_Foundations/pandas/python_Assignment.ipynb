```python
filtered_data = df1[
    (df1['petal_length'] > 4.5) & (df1['species'] == 'Iris-virginica')
]

filtered_data
```

```python
grouped_iris = df1.groupby('species').agg({
    'sepal_length': 'mean',
    'petal_width' : 'max',
    'sepal_width' : 'std'
})
grouped_iris
```

```python
#Feature Engineering Creating new column out of existing ones
df1['petal_ratio'] = df1['petal_length'] / df1['petal_width']

avg_ratio = df1.groupby("species")["petal_ratio"].mean()
avg_ratio
```
