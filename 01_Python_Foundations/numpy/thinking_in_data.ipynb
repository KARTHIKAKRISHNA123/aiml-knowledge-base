```python
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_users = []
    unique_users = set()
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

    

        #Handling missing data
        raw_age = user.get("age")
        if (raw_age == None):
            user["age"] = None
    
        #Deduplication
        if user["name"] not in unique_users:
            unique_users.add(user["name"])
            cleaned_users.append(user)
    
    return cleaned_users

    
```

```python
data = clean_data(data)
data
```

```python
#Get Meaningful insights from data
def get_insights(data):
    #Average Rating and percentage of users with poor data
    total_rating = 0
    poor_rating = 0
    for user in data:
        total_rating += float(user["rating"])
        if (float(user["rating"]) < 3):
            poor_rating += 1
        
    
    avg_rating = total_rating / len(data)
    poor_rating_percent = poor_rating / len(data) * 100
    
    print(f"Average Rating = {avg_rating}")
    print(f"Percentage of users with poor rating = {poor_rating_percent} %")
```

```python
get_insights(data)
```

```python
#Build Recommendation Feature
def get_recommendations(data):
    recommendations = []
    
    for user in data:
        curr_recom = {}
        curr_recom["name"] = user["name"]
        if (float(user["rating"]) >= 4):
            curr_recom["brand"] = "Apple"
        else:
            curr_recom["brand"] = "Samsung"
        recommendations.append(curr_recom)
    print(f"The Recommendation list for users is {recommendations}")
            
    
```

```python
get_recommendations(data)
```
