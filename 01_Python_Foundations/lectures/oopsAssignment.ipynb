```python
class BankAccount:

    def __init__(self, account_number, owner_name, balance):
        self.account_number = account_number
        self.owner_name = owner_name
        self.balance = balance
        print(f"The Account created with {self.account_number} for the account owner {self.owner_name} with balance {self.balance}.")

    def deposit(self, amount):
        if amount < 0:
            print("Invalid Amount Entered to deposit")
        else:
            self.balance = amount + self.balance
            print(f"The deposited amount is {amount}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient Balance cannot withdraw this amount")
        elif amount < 0:
            print("Invalid amount to withdraw")
        else:
            self.balance = self.balance - amount
            print(f"The amount withdrawn is {amount}.")

    def checkbalance(self):
        print(f"The available balance is {self.balance}")


KK_Account = BankAccount(12, "KK", 10000000)
KK_Account.deposit(23000)
KK_Account.withdraw(2000)
KK_Account.checkbalance()
```

```python
class Book:
    def __init__(self, book, author):
        self.book = book
        self.author = author
        self.reviews = []

    def add_review(self, review_text):
        self.reviews.append(review_text)
        print("Review Appended")

    def count_reviews(self):
        print(len(self.reviews))

    def display_reviews(self):
        if len(self.reviews) == 0:
            print("No Reviews yet")
        else:
            for item in self.reviews:
                print(item)

my_book = Book("Reo", "KK")
my_book.add_review("fantastic")
my_book.add_review("Super")
my_book.add_review("Wonderful")
my_book.count_reviews()
my_book.display_reviews()

```

```python
import math
class Shape:
    
    def area(self):
        print("Area of generic shapes")
        return 0

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        
        return math.pi * self.radius * self.radius

class Rect(Shape):

    def __init__(self, radius1, radius2):
        self.radius1 = radius1
        self.radius2 = radius2
    def area(self):
        return self.radius1 * self.radius2

class Tri(Shape):
    def __init__(self, height, width):
        self.height = height
        self.width = width

    def area(self):
        return 0.5 * self.height * self.width


shape = [Shape(), Circle(5), Rect(3, 4), Tri(3, 4)]
for s in shape:
    print(s.area())
```

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def show_details(self):
        print(f"The vehicle model is {self.model} and its brand is {self.brand}")

class Car(Vehicle):
    def __init__(self, brand, model, seats):
        super().__init__(brand, model)
        self.seats = seats

    def show_details(self):
        print(f"The Car model is {self.model} and its brand is {self.brand}, it has {self.seats} number of seats")

class Bike(Vehicle):
    def __init__(self, brand, model, engine):
        super().__init__(brand, model)
        self.engine = engine

    def show_details(self):
        print(f"The Car model is {self.model} and its brand is {self.brand}, it has {self.engine} cc engine")


vehicles = [Vehicle("xyz", "xyz"), Car("Toyoto", "xyz", 5), Bike("Yamaha", "xyz", 150)]

for item in vehicles:
    print(item.show_details())
        
    
```

```python
from abc import ABC, abstractmethod

class Employee:
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def calculate_self(self):
        pass

class Intern(Employee):
    def __init__(self, name):
        super().__init__(name)
        
    def calculate_salary(self):
        return 5000

class Contract_employee(Employee):
    def __init__(self, name, hours_worked, hourly_rate):
        super().__init__(name)
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class Full_Time_Employee(Employee):
    def __init__(self, name):
        super().__init__(name)
        
    def calculate_salary(self):
        return 50000

staff = [Intern("KK"), Full_Time_Employee("JJ"), Contract_employee("Dev", 10, 1000)]

for emp in staff:
    print(emp.name)
    print(emp.calculate_salary())
        
    
        
```

```python
class Player:
    player_count = 0
    def __init__(self, name, level):
        self.name = name
        self.level = level
        Player.player_count += 1

    @classmethod
    def get_total_players(cls):
        return cls.player_count


p1 = Player("KK", 1)
p2 = Player("JJ", 2)

print(Player.get_total_players(), p1.name, p1.level)

        
```

```python
class Herbivore:
    def eat(self):
        print("Eats Leaves and Berries")

class Omnivore:
    def eat2(self):
        print("I eat all")

class Carnivore:
    def eat1(self):
        print("I eat only meals")

class Bear(Herbivore, Omnivore, Carnivore):
    def roar(self):
        print("I can also roar")

pooh = Bear()
pooh.eat()
pooh.eat1()
pooh.eat2()
pooh.roar()
```

```python
class Message:
    counter = 1
    def __init__(self, sender, content):
        self.id = Message.counter
        Message.counter += 1
        self.sender = sender
        self.content = content

    def __str__(self):
        return f"[{self.id}] {self.sender}: {self.content}"

class User:
    def __init__(self, username):
        self.username = username
        self.room = None

    def join_room(self, chatroom):
        self.room = chatroom
        chatroom.add_user(self)

    def send(self, text):
        if self.room:
            self.room.broadcast(self, text)

        else:
            print("Enter a room first")

class ChatRoom:
    def __init__(self, name):
        self.name = name
        self.users = []
        self.messages = []

    def add_user(self, user):
        self.users.append(user)
        print(f"{user.username} joined {self.name}")

    def broadcast(self, sender, content):
        new_msg = Message(sender.username, content)
        self.messages.append(new_msg)
        print(new_msg)

    def show_history(self):
        print(f"History of {self.name} is:")
        for m in self.messages:
            print(m)

room = ChatRoom("dev Team")
u1 = User("Alice")
u2 = User("Bob")

u1.join_room(room)
u2.join_room(room)

u1.send("Hi Iam an SDE")
u2.send("Hi! Same to you")
u1.send("Hi Iam an SDE")
u2.send("Hi! Same to you")


room.show_history()
```
