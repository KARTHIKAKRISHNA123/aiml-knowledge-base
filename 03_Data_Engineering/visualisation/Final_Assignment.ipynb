```python
tips = sns.load_dataset("tips")
tips.head()
```

```python

fig, ax = plt.subplots()
sns.boxplot(
    data=tips,
    x="total_bill",
    y="day",
    hue="sex",
    ax=ax
)

fig.tight_layout()
```

```python
fig, ax = plt.subplots()
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    ax=ax,
    
)

ax.set_title("Total Bill Vs Tip")
ax.set_xlabel("Total Bill")
ax.set_ylabel("Tip")

fig.tight_layout()

```

```python
import pandas as pd

pivot_table = tips.pivot_table(index="day", columns="time", values="tip", observed=False)

```

```python
pivot_table
```

```python
sns.heatmap(
    pivot_table,
    annot = True,
    cmap="coolwarm",
    fmt=".2f"
)
    
    
```
