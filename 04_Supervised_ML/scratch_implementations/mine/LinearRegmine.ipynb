```python
import numpy as np
```

```python
#Linear Regression
class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def fit(self, x, y): #x_train, y_train
        m, n = x.shape #(number of samples, number of features) (rows, columns)

        #step 1: initialize parameters
        self.bias = 0
        self.weights = np.zeros(n)

        for i in range(self.n_iter): #Gradient Descent


            #step 2: gradient descent calculation or y_pred calc
            y_pred = self.bias + np.dot(x, self.weights)

            #step 3: calc gradients
            db = (1/m) * np.sum(y_pred - y) #derivative of bias
            dw = (1/m) * np.dot(x.T, (y_pred - y)) #derivative of weights

            #step 4: convergence theorem - params update
            self.bias = self.bias - self.lr * db
            self.weights = self.weights - self.lr * dw




    def predict(self, x):
        y_pred = self.bias + np.dot(x, self.weights)
        return y_pred


```

```python
x = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(x, y)

y_pred = model.predict(x)
y_pred
```

```python
#Linear Regression with OLS

class LinearRegressionOLS:
    def __init__(self):
        self.bias = None
        self.weights = None


    def fit(self, x, y):
        m, n = x.shape

        x_b = np.c_[np.ones((m, 1)), x]

        #Normal Equation
        #theta = np.dot(np.linealg.inv(np.dot(x_b.T, x_b)), np.dot(x_b.T, y))

        theta = np.linalg.inv(x_b.T @ x_b) @ x_b.T @ y

        self.bias = theta[0]
        self.weights = theta[1: ]




    
    def predict(self, x):
        y_pred = self.bias + np.dot(x, self.weights)
        return y_pred
```

```python
x = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegressionOLS()
model.fit(x, y)

y_pred = model.predict(x)
y_pred
```
