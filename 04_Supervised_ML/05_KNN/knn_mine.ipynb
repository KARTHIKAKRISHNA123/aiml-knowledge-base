```python
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
heart_df = pd.read_csv("1-heart.csv")

x = heart_df.drop("target", axis=1)
y = heart_df["target"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
#Scaling the data so tahat all the features are on the same scale and the model can learn better
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled
```

```python
knn_classifier = KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(x_train_scaled, y_train)
```

```python
y_pred = knn_classifier.predict(x_test_scaled)




```

```python
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))


```

```python
knn_classifier = KNeighborsClassifier(n_neighbors=5)
knn_classifier.fit(x_train_scaled, y_train)
```

```python
y_pred = knn_classifier.predict(x_test_scaled)




```

```python
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))


```

```python
knn_classifier = KNeighborsClassifier(n_neighbors=7) #best score till now
knn_classifier.fit(x_train_scaled, y_train)
```

```python
y_pred = knn_classifier.predict(x_test_scaled)




```

```python
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))


```

```python
#Cross Validation for hyper parameter tuning using GridSearchCV
from sklearn.model_selection import GridSearchCV
```

```python
classifier = KNeighborsClassifier()
param_grid = {
    "n_neighbors": [3, 5, 7, 9]

}

classifierCV = GridSearchCV(
    classifier,
    param_grid,
    cv=None,
)

classifierCV.fit(x_train_scaled, y_train)

y_pred = classifierCV.predict(x_test_scaled)

print("Recall Score: ", recall_score(y_test, y_pred))
print("Accuracy Score: ", accuracy_score(y_test, y_pred))
print("Precision Score: ", precision_score(y_test, y_pred))
```

```python
#results
res = pd.DataFrame(classifierCV.cv_results_)
```

```python
#KNN with Pipeline
from sklearn.pipeline import Pipeline

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

steps = [
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier())
]

pipeline= Pipeline(steps)

param_grid = {
    "knn__n_neighbors": [3, 5, 7, 9]

}

classifierCV = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="recall"
)

classifierCV.fit(x_train, y_train)

y_pred = classifierCV.predict(x_test)

print("Recall Score: ", recall_score(y_test, y_pred))
print("Accuracy Score: ", accuracy_score(y_test, y_pred))
print("Precision Score: ", precision_score(y_test, y_pred))

```
