```python
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import pandas as pd
```

```python
iris = load_iris()

```

```python
iris
```

```python
X = iris.data

y = iris.target
```

```python
y
```

```python
X.shape
```

```python
#Visualization
sns.scatterplot(x=X[:, 0], y=X[:, 1])
```

```python
#Visualization
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=y)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
# Optional: Dimensionality Reduction using PCA for better visualization 
from sklearn.decomposition import PCA
pca = PCA(n_components=2) 
X_pca = pca.fit_transform(X_scaled)
```

```python
# Visualize the data
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=y)
```

```python
# Elbow Method to find optimal K
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_pca)
    wcss.append(kmeans.inertia_)

    
```

```python
wcss
```

```python
sns.lineplot(x=range(1, 11), y=wcss, marker="o")
```

```python
# Kmeans with K=3
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X_pca)

sns.scatterplot(x=X[:, 0], y=X[:, 2], hue=labels)

```

```python
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=labels)
sns.scatterplot(x=kmeans.cluster_centers_[:, 0], y=kmeans.cluster_centers_[:, 1], s=200, color="black", marker="X")
```
