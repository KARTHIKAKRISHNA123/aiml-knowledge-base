```python
import json

cities = {
    "New York": 100000,
    "Pennsylvania": 20000,
    "San Francisco": 30000
}

# Write initial data
with open("cities.json", "w") as f:
    json.dump(cities, f, indent=4)

# Read data
with open("cities.json", "r") as f:
    loaded_data = json.load(f)
    for city, pop in loaded_data.items():
        print(f"{city}: {pop}")

# Add new city
n = input("Enter a New City: ")
p = int(input(f"Enter Population of {n}: "))

loaded_data[n] = p

# Write updated data
with open("cities.json", "w") as f:
    json.dump(loaded_data, f, indent=4)

# Read again
with open("cities.json", "r") as f:
    print("After Modification")
    loaded_data2 = json.load(f)
    for city, pop in loaded_data2.items():
        print(f"{city}: {pop}")
```
