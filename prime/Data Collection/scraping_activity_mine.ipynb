```python
#Fetch the data and store it for testing
import requests
from bs4 import BeautifulSoup
import time # Good practice to import this

url = "https://quotes.toscrape.com/page/1/"
page_count = 1

while True:
    print(f"Attempting to fetch page {page_count}...") # debug print

    try:
        # Added timeout to prevent hanging forever
        res = requests.get(url, timeout=10) 
        
        soup = BeautifulSoup(res.text, "lxml")
        quotes = soup.select("div.quote")

        if not quotes:
            print(f"No quotes found on page {page_count}. Stopping.")
            break

        # Ensure the directory exists before writing (optional safety)
        with open(f"scraped_data/quotes{page_count}.html", "w", encoding='utf-8') as f:
            f.write(res.text)
        
        print(f"Data scraped successfully from {page_count}")
        
        page_count += 1
        url = f"https://quotes.toscrape.com/page/{page_count}/"
        
        # Be polite to the server (prevents getting banned)
        time.sleep(1) 

    except requests.exceptions.RequestException as e:
        print(f"Network error on page {page_count}: {e}")
        break
```

```python
#Extract useful info from the above data - page 1
life_quotes = []
page_count = 1
while True:

    if page_count == 11:
        break
    with open(f"scraped_data/quotes{page_count}.html", "r", encoding="utf-8") as f:
        html_content = f.read()

        
    
    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")
    
    


    for q in all_quotes:
        all_tags = []
        

        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())

        if "life" in all_tags:
            print(f"from {page_count}")
            print()
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes.append([text, author])
    print(life_quotes)
             
            


    page_count = page_count + 1;
```

```python
print(len(life_quotes))
for quotes in life_quotes:
    print(quotes)
```

```python
import pandas as pd

df = pd.DataFrame(life_quotes, columns = ["Quote", "Author"])
df.head()
df.to_csv("cleaned_data/life_quotes.csv", index=False)
```
