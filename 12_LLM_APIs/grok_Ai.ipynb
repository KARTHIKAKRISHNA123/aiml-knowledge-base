```python
import os, time
from google import genai
from google.genai import errors

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))
MODELS = ["gemini-3.8-flash", "gemini-2.5-flash"]

def call_with_fallback(make_call, retries=4):
    last_error = None
    for model in MODELS:
        for attempt in range(retries):
            try:
                return make_call(model)
            except errors.ServerError as e:
                last_error = e
                time.sleep(2 ** attempt)
    raise last_error
```

```python
import httpx
from google.genai import types

# Image
image_url = "https://images.pexels.com/photos/3476860/pexels-photo-3476860.jpeg"  # use your full URL
image_bytes = httpx.get(image_url, follow_redirects=True).content

response = call_with_fallback(
    lambda model: client.models.generate_content(
        model=model,
        contents=[
            types.Part.from_bytes(data=image_bytes, mime_type="image/jpeg"),
            "QUERY",   "Describe this image"
        ],
    )
)

print(response.text)
```

```python

from google import genai
from google.genai import types
import os

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

response = client.models.generate_content(
    model="gemini-2.5-flash-image",
    contents=(
        "Generate an image of a highly detailed futuristic "
        "spaceship, flying near a nebula"
    ),
    config=types.GenerateContentConfig(
        response_modalities=["IMAGE", "TEXT"]
    )
)

# Extract generated image data
image_data = [
    part.inline_data.data
    for part in response.candidates[0].content.parts
    if part.inline_data is not None
]

# Decode and store image in a file
if image_data:
    image_bytes = image_data[0]

    with open("spaceship.png", "wb") as f:
        f.write(image_bytes)

    print("Image saved as spaceship.png")
else:
    print("No image was generated.")
```

### Images API

```python

import os
from google import genai
from google.genai import types

prompt = (
    "Generate an image of a highly detailed futuristic "
    "spaceship, flying near a nebula"
)

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

response = client.models.generate_content(
    model="gemini-2.5-flash-image",
    contents=prompt,
    config=types.GenerateContentConfig(
        response_modalities=["IMAGE", "TEXT"]
    )
)

# Extract and save the generated image
for part in response.candidates[0].content.parts:
    if part.inline_data is not None:
        image = part.as_image()
        image.save("spaceship-image.png")
        print("Image saved as spaceship-image.png")
        break
else:
    print("No image was generated.")
```

```python

import os
import base64
from google import genai

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

# EDIT: Change the spaceship color to red
prompt = (
    "Take the reference image and change only the spaceship "
    "color to red. Preserve its shape, details, lighting, "
    "background, and composition."
)

# Read the reference image
with open("spaceship-image.png", "rb") as f:
    image_bytes = f.read()

# Send the prompt and reference image to Gemini
response = client.interactions.create(
    model="gemini-3.1-flash-image",
    input=[
        {
            "type": "text",
            "text": prompt
        },
        {
            "type": "image",
            "mime_type": "image/png",
            "data": base64.b64encode(image_bytes).decode("utf-8")
        }
    ]
)

# Extract and save the edited image
for step in response.steps:
    if step.type == "model_output":
        for content in step.content:
            if content.type == "text":
                print(content.text)

            elif content.type == "image":
                with open("spaceship-red.png", "wb") as f:
                    f.write(base64.b64decode(content.data))

                print("Edited image saved as spaceship-red.png")
```
