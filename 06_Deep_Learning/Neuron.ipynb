```python
import torch 
import torch.nn as nn
```

```python
torch.manual_seed(42)
inputs = torch.tensor([1.0, 2.0, 3.0])
```

```python
neuron = nn.Linear(in_features=3, out_features=1)
output = neuron(inputs)
```

```python
neuron.bias
```

```python
neuron.weight
```

```python
output == neuron.weight @ inputs +neuron.bias
```
